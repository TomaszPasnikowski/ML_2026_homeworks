# Transposed convolutions

Transposed convolutions are the building blocks of the decoder in convolutional autoencoder. They mirror the classical convolution layers in the encoder in order to reverse their operations.

While convolutions downscale and reduce the size of input, transposed convolutions upscale data into a larger size. The idea is simple

- In convolutions many input pixels create one output pixel through kernel.
- In transposed convolutions one input pixel affects many output pixel through kernel.

In transposed convolution each input pixel is multiplied by the whole kernel and when there is overlapping between outputs of different pixels, we just sum the values (examples below). The resulting image has therefore larger demension.

Similarly to classical convolutions, transposed convolutions may have different kernel sizes, paddings and strides.

- Kernel size determines how many output pixels will each input pixel affect. Bigger kernel means larger output resolution and also larger context size of the model (possibility to learn bigger patterns).
- Stride works analogously as in classical convolutions. It sets the spaces between outputs of the pixels. Bigger stride means larger output resolution and less overlapping between the pixels' output areas.
- Padding in transposed convolutions removes the areas that are affected by the least amount of input pixels - the outer coordinates of the output. For example padding of size 2 means that after computing the result the first and last two rows will be discarded as well as first and last two columns. Larger padding means smaller output dimension.
